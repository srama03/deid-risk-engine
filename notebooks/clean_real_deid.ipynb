```python
import pandas as pd
```

```python

# cleaning data
df = pd.read_csv("/content/ncvoter60.csv", sep="\t", encoding="ISO-8859-1")
```

```python
df.head()
```

```python
df.shape
```

```python
# checking if decoded correctly
df[df["last_name"]=="O"]
```

```python
# remove "removed"/"denied" voter desc
df = df[df['voter_status_desc'].isin(['ACTIVE', 'INACTIVE'])]
print(df.shape)
```

```python
df['precinct_abbrv'].isna().sum()
df = df.dropna(subset=["precinct_abbrv"])
print(df.shape)
```

```python
# Columns we need to keep
columns_to_keep = [
    # Names
    "last_name",
    "first_name",
    "middle_name",
    # Address
    "res_street_address",
    "res_city_desc",
    "precinct_abbrv",
    # Demographics
    "race_code",
    "ethnic_code",
    "party_cd",
    "gender_code",
    "birth_year",
    "age_at_year_end",
]

df = df[columns_to_keep].copy()
print(df.shape)
df.head(6)
```

```python
df[df.columns].isna().sum()
```

```python
# drop fn/ln nas
df = df.dropna(subset=["last_name", "first_name", "res_city_desc"])
print(df.shape)
```

```python
pres = df.groupby("precinct_abbrv").size()
pres.sort_values().head(10)
```

```python
# shuffle zip entries
pres = pres.sample(frac=1.0, random_state=42)
print(pres.cumsum())

```

```python
shifted = pres.cumsum().shift(1)
condition = (shifted < 10000) | (shifted.isna())
pres[condition]
```

```python
idx = pres[condition].index
df_scoped = df[df["precinct_abbrv"].isin(idx)]
print(df_scoped.shape)
```

```python
df_scoped["precinct_abbrv"].unique()
```

```python
df["res_street_address"] = df["res_street_address"].str.strip()
```

```python
df_scoped.to_csv("ncvoter_final.csv", index=False)
```
